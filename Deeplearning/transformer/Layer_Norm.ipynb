```python
import numpy as np
```

```python
def layer_normalization(X: np.ndarray, gamma: np.ndarray, beta: np.ndarray, epsilon: float = 1e-5) -> np.ndarray:

    # compute mean and var for each row
    mu = np.mean(X, axis = -1, keepdims =True)
    var= np.var(X, axis = -1, keepdims =True)

    # Normalize X 
    X_scaled = (X - mu) / (np.sqrt(var + epsilon))

    # Scale X
    y = gamma * X_scaled + beta

    return y     
```

```python
np.random.seed(42)
X = np.random.randn(2, 2, 3)
gamma = np.ones(3).reshape(1, 1, -1)
beta = np.zeros(3).reshape(1, 1, -1)
epsilon = 1e-5
```

```python
layer_normalization(X, gamma, beta, epsilon)
```
