```python
import numpy as np
```

```python
def pos_encoding(position:int, dmodel:int):
    
    if position == 0 or dmodel == 0: return -1

    pos = np.array(np.arange(position), np.float16)
    ind = np.array(np.arange(dmodel), np.float16)
    pos = pos[:, np.newaxis] # (pos, 1)
    ind = ind[np.newaxis,:]  # (1, dmodel)
    #print(pos)
    #print(ind)

    # Compute angle rates
    angle_rates = 1 / np.power(10000, (2 * (ind // 2)) / np.float32(dmodel))

    # Compute the positional encoding matrix
    angle_rads = pos * angle_rates  # ( pos, dmodel)
    #print(angle_rads.shape)

    # apply sine to even indices and cosine to odd indices
    # Apply sin to even indices in the array; cos to odd indices
    angle_rads[:, 0::2] = np.sin(angle_rads[:, 0::2])  # even index (0, 2, 4, ...)
    angle_rads[:, 1::2] = np.cos(angle_rads[:, 1::2])  # odd index (1, 3, 5, ...)

    return angle_rads.astype(np.float16)
```
