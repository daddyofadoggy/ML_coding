```python
import numpy as np
```

```python
def topk(H: np.ndarray, k: int) -> np.ndarray:
        """
        For each row in matrix H, keeps the top-k values and sets the rest to -inf.
    
        Args:
            H (np.ndarray): Input 2D array of shape (batch_size, num_experts)
            k (int): Number of top elements to keep per row
    
        Returns:
            np.ndarray: Same shape as H, with only top-k elements kept and rest set to -inf
        """
        # Copy H to avoid modifying the original
        H_topk = np.full_like(H, -np.inf)  # Initialize with -inf
    
        # Get indices of top-k elements in each row
        topk_indices = np.argpartition(-H, kth=k-1, axis=1)[:, :k]  # partial sort
    
        # Assign top-k values to their positions
        for i in range(H.shape[0]):
            H_topk[i, topk_indices[i]] = H[i, topk_indices[i]]
    
        return H_topk
```

```python
def noisy_topk_gating(
    X: np.ndarray,
    W_g: np.ndarray,
    W_noise: np.ndarray,
    N: np.ndarray,
    k: int
) -> np.ndarray:
    
    H_base = X @ W_g
    H_noise = X @ W_noise
    softplus = np.log1p(np.exp(H_noise))
    H = H_base + N * softplus
    masked_H = topk(H,k)
    exp = np.exp( masked_H - np.max(masked_H, axis = 1, keepdims = True))
    return exp / np.sum(exp)
```
