```python
#!/usr/bin/env python3
"""
LaTeX to PNG Renderer
Renders LaTeX equations to high-resolution PNG images using matplotlib
"""

import matplotlib.pyplot as plt
import matplotlib as mpl
from matplotlib import mathtext
import numpy as np

def render_latex_to_png(latex_string, filename="latex_equations_200dpi.png", dpi=200):
    """
    Render LaTeX equations to PNG with specified DPI
    
    Args:
        latex_string (str): LaTeX equation string
        filename (str): Output filename
        dpi (int): Resolution in dots per inch
    """
    
    # Configure matplotlib for high-quality rendering
    plt.rcParams['text.usetex'] = False  # Use mathtext instead of full LaTeX
    plt.rcParams['font.size'] = 16
    plt.rcParams['mathtext.fontset'] = 'cm'  # Computer Modern font
    
    # Create figure with white background
    fig, ax = plt.subplots(figsize=(8, 3))
    fig.patch.set_facecolor('white')
    ax.set_facecolor('white')
    
    # Hide axes
    ax.axis('off')
    
    # Your LaTeX equations
    latex_text = r'''$ G = \text{Softmax}(H') $'''
    
    # Render the LaTeX
    ax.text(0.5, 0.5, latex_text, 
            horizontalalignment='center',
            verticalalignment='center',
            transform=ax.transAxes,
            fontsize=20,
            bbox=dict(boxstyle="round,pad=0.3", facecolor="white", edgecolor="none"))
    
    # Adjust layout to remove extra whitespace
    plt.tight_layout()
    
    # Save with high DPI
    plt.savefig(filename, dpi=dpi, bbox_inches='tight', 
                facecolor='white', edgecolor='none', format='png')
    
    print(f"✅ LaTeX equations saved to: {filename}")
    print(f"📊 Resolution: {dpi} DPI")
    
    # Show the plot
    plt.show()
    
    plt.close()

def render_with_alternative_method(filename="latex_equations_alt_200dpi.png", dpi=200):
    """
    Alternative rendering method using individual equations
    """
    
    # Create figure
    fig, ax = plt.subplots(figsize=(6, 2))
    fig.patch.set_facecolor('white')
    ax.set_facecolor('white')
    ax.axis('off')
    
    # Render equations separately for better control
    eq1 = r'$ H_{\text{base}} = X W_g $'
    eq2 = r'$ H_{\text{noise}} = X W_{\text{noise}} $'
    
    ax.text(0.5, 0.7, eq1, horizontalalignment='center', 
            transform=ax.transAxes, fontsize=18)
    ax.text(0.5, 0.3, eq2, horizontalalignment='center', 
            transform=ax.transAxes, fontsize=18)
    
    plt.tight_layout()
    plt.savefig(filename, dpi=dpi, bbox_inches='tight', 
                facecolor='white', edgecolor='none', format='png')
    
    print(f"✅ Alternative version saved to: {filename}")
    plt.show()
    plt.close()
```
