```python
import numpy as np
class Kmeans:
    def __init__(self, k=3, iteration =1000, tol=1e-4 ):
        self.k = k
        self.iteration = iteration
        self.tol = tol

    def fit(self, X):
        # random init centroids
        self.centroids = X[np.random.choice(range(len(X)), self.k)]
        cluster_assign = np.zeros(len(X), dtype=int)

        for i in range(self.iteration):
            # find the closest distance centroids
            for j in range(len(X)):
                distances =  np.linalg.norm(X[j] - self.centroids, axis = 1)
                cluster_assign[j] = np.argmin(distances)
            
            # update the centroids
            for m in range(self.k):
                points = X[np.where(np.array(cluster_assign) == m)]
                
                if len(points)>0:
                    self.centroids[m] = np.mean(points, axis = 0)
            
            if np.allclose(prev_centroids, self.centroids, atol = self.tol):
                print(f"Converged after {iteration + 1} iterations")
                break

            prev_centroids = np.copy(self.centroids)
    
        self.cluster_assign = cluster_assign
    
    def predict(self, X):
        cluster_assign = np.zeros(len(X), dtype=int)
        for j in range(len(X)):
            distances =  np.linalg.norm(X[j] - self.centroids, axis = 1)
            cluster_assign[j] = np.argmin(distances)
        return cluster_assign
        

```
