```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Init weights
w1 = np.linspace(-3, 3, 400)
w2 = np.linspace(-3, 3, 400)

# Create a meshgrid
W1, W2 = np.meshgrid(w1, w2)
```

```python
def elasticnet(alpha = 0.5, W1 =0 , W2 = 0):
    return alpha * (np.abs(W1) + np.abs(W2)) + (1 - alpha) * (W1 ** 2 + W2 ** 2)
```

```python
# Define a simple quadratic loss function (e.g., least squares)
loss = W1 ** 2 + 2 * W2 ** 2  # Elliptical contours
```

```python
# l1 constraint 
l1_constraint = np.abs(W1) + np.abs(W2)

# l2 constraint
l2_constraint = W1 ** 2 + W2 ** 2

# elasticnet constraint
elasticnet_constraint = elasticnet(0.5, W1, W2)
```

```python
# Plot 
fig, axs = plt.subplots(1,3, figsize=(18, 5))

# plot the L1 contour
axs[0].contour(W1, W2, loss, levels = 20, cmap = 'Blues')
axs[0].contour(W1, W2, l1_constraint , levels = [1], colors = 'Red')
axs[0].set_title('L1 - Regularization')
axs[0].set_xlabel('w1')
axs[0].set_ylabel('w2')
axs[0].axhline(0, color = 'grey', lw =0.5)
axs[0].axvline(0, color = 'grey', lw =0.5)

# plot the L2 contour
axs[1].contour(W1, W2, loss, levels = 20, cmap = 'Blues')
axs[1].contour(W1, W2, l2_constraint , levels = [1], colors = 'Red')
axs[1].set_title('L2 - Regularization')
axs[1].set_xlabel('w1')
axs[1].set_ylabel('w2')
axs[1].axhline(0, color = 'grey', lw =0.5)
axs[1].axvline(0, color = 'grey', lw =0.5)

# plot the elasticnet contour
axs[2].contour(W1, W2, loss, levels = 20, cmap = 'Blues')
axs[2].contour(W1, W2, elasticnet_constraint , levels = [1], colors = 'Red')
axs[2].set_title('Elasticnet - Regularization (alpha = 0.5)')
axs[2].set_xlabel('w1')
axs[2].set_ylabel('w2')
axs[2].axhline(0, color = 'grey', lw =0.5)
axs[2].axvline(0, color = 'grey', lw =0.5)
```
