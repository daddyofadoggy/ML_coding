```python
class NaiveBayes():
    def __init__(self, smoothing=1.0):
        # Initialize smoothing
        #pass
        self.smoothing = smoothing
        #self.classes_ = np.array([0, 1])
        self.single_class = None # flag to check if only one class exists

    def forward(self, X, y):
        # Fit model to binary features X and labels y
        #pass
        nemails, nwords = X.shape
        unique_classes = np.unique(y)
        if len(unique_classes)==1:
            self.single_class = unique_classes[0]
            return self
        # Prior
        self.classes_ = np.unique(y)
        n_classes = np.bincount(y) 
        self.log_prior = np.log(n_classes/nemails)

        # Word Probability for each class
        word_present = np.array([X[y == 0].sum(axis = 0),
                X[y == 1].sum(axis = 0)])
        
        emails_per_class = n_classes.reshape(-1, 1) # vector to array
        # Laplace smoothing
        theta = (word_present + self.smoothing) / (emails_per_class + 2 * self.smoothing) 

        self.log_theta_present_ = np.log(theta)      # when word flag = 1
        self.log_theta_absent_  = np.log(1 - theta)  # when word flag = 0
        return self

    def predict(self, X):
        # Predict class labels for test set X
        # pass
        if self.single_class is not None:
            # Return only that class for all predictions
            return np.full(X.shape[0], self.single_class)

        score = self.log_prior + X @ self.log_theta_present_.T + (1 - X) @ self.log_theta_absent_.T
        return self.classes_[score.argmax(axis =1)]
```
