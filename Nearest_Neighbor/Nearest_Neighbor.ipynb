# 🧭 When to Use Euclidean, Manhattan, or Minkowski Distance in Nearest Neighbor Algorithms

Nearest neighbor algorithms like **KNN** rely on a distance metric to determine similarity. The three most common are **Euclidean**, **Manhattan**, and **Minkowski**. Choosing the right one depends on your data and the problem context.

---

## ✅ 1. Euclidean Distance (Minkowski with *p* = 2)

**Formula**:  
\[
D(x, y) = \sqrt{\sum (x_i - y_i)^2}
\]

### Use When:
- Features are **continuous** and **equally scaled** (or standardized).
- You expect the shortest "as-the-crow-flies" distance to matter.
- Data clusters are **spherical** in feature space.

### Good For:
- Image classification  
- Physical/real-world spatial data  
- Classic ML datasets like Iris

---

## ✅ 2. Manhattan Distance (Minkowski with *p* = 1)

**Formula**:  
\[
D(x, y) = \sum |x_i - y_i|
\]

### Use When:
- Data is **sparse** or **high-dimensional**.
- You care more about **accumulated differences**.
- Movement is restricted to a **grid**, like navigating city blocks.

### Good For:
- Text classification (bag-of-words, one-hot encoding)  
- Pathfinding in grid environments  
- High-dimensional binary or categorical features

---

## ✅ 3. Minkowski Distance (Generalized, *p* ≥ 1)

**Formula**:  
\[
D(x, y) = \left( \sum |x_i - y_i|^p \right)^{1/p}
\]

### Use When:
- You want to **tune the distance behavior** using the parameter *p*.
- Not sure if Euclidean or Manhattan works better — try both and validate.
- Larger *p* values emphasize large individual differences between features.

### Good For:
- Custom distance needs  
- Feature importance tuning  
- Optimization experiments (e.g., via cross-validation)

---

## 🧪 Quick Heuristic

| **Data Type**                | **Recommended Distance** |
|-----------------------------|---------------------------|
| Continuous, normalized      | Euclidean                 |
| Sparse or high-dimensional  | Manhattan                 |
| Tuning/custom similarity    | Minkowski (vary *p*)      |

---

## ⚠️ Important Reminder:
Always **scale your features** (e.g., with StandardScaler) before applying any distance-based algorithm, or features with larger magnitudes will dominate the distance calculation.
C

```python
class KNN:
    def __init__(self, k, distance = 'euclidian'):
        self.k = k
        self.distance = distance
        
    def fit(self, X, y):
        self.X_train = X
        self.y_train = y

    def predict(self, X_test):
        y_pred = []
        for x in X_test:
            # Compute the distance of X_text (single instance ) to all training data
            if self.distance == 'euclidian':
                distances = np.linalg.norm (self.X_train - x , axis = 1)
            else:
                # 'manhattan' distance
                distances = np.sum ( np.abs(self.X_train - x), axis = 1)

            ## Find KNN and classes 
            knnindices = np.argsort ( distances )[ : self.k]
            classes =  self.y_train[knnindices]
            
            ## Assign label by majority voting 
            label = Counter(classes).most_common(1)[0][0]
            y_pred.append(label)
            
        return np.array(y_pred)
```

## TEST ##

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# Load the iris dataset
iris = load_iris()

# Split the data into training and test sets
X_train, X_test, y_train, y_test = train_test_split(iris.data, iris.target, test_size=0.2, random_state=42)

# Create a KNN classifier with k=5 and euclidean distance
knn = KNN(k=5, distance='euclidean')

# Train the classifier on the training data
knn.fit(X_train, y_train)

# Make predictions on the test data
y_pred = knn.predict(X_test)

# Compute the accuracy of the classifier
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy}")
```
